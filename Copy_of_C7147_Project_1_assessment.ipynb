## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """
    Normalises a sequencing read to uppercase and validates its length
    and nucleotide composition against a reference sequence.

    Parameters:
        sequence (str): The nucleotide sequence of the read to validate.
        reference (str): The reference sequence (uppercase A, C, G, T).

    Returns:
        dict: A dictionary containing:
            - 'sequence' (str): The uppercase version of the input sequence.
            - 'valid' (bool): True if the read has no errors, False otherwise.
            - 'errors' (list): List of error strings ('wrong_length',
              'invalid_characters') in required order.
    """
    # Write your code here.
#print(reference)

    upper_sequence = sequence.upper() #make sequence uppercase
    errors = []

    #check for length matching reference sequence
    if len(upper_sequence) != len(reference):
      errors.append("wrong_length")

    #every base must be A, C, T, or G
    valid_bases = {"A", "C", "T", "G"}
    if not set(upper_sequence).issubset(valid_bases):
      errors.append("invalid_characters")

    #read valid if no errors are recorded
    is_valid = len(errors) == 0


    return {
        "sequence": upper_sequence,
        "valid": is_valid,
        "errors": errors
    }



````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """
    Identifies all nucleotide mismatches between a validated sequencing read
    and a reference sequence.

    Parameters:
        sequence (str): A valid uppercase nucleotide read sequence.
        reference (str): The uppercase reference sequence of the same length.

    Returns:
        list: A list of dictionaries, one per mismatch, ordered by ascending
              1-based position. Each dictionary contains:
              - 'position' (int): 1-based nucleotide coordinate.
              - 'reference_base' (str): Nucleotide in the reference sequence.
              - 'read_base' (str): Mismatched nucleotide in the sequencing read.
    """
    # Write your code here.

    mutations = []

    #paired bases from reference and read starting from position 1
    for position, (ref_base, read_base) in enumerate(zip(reference, sequence), start = 1):
      if read_base != ref_base:
        mutations.append({
            "position": position,
            "reference_base": ref_base,
            "read_base": read_base
        })

    return mutations
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
  """
  Validates a dataset of sequencing reads and identifies mutations for all
  valid reads against a reference sequence, without modifying the input data.

  Parameters:
      reads (list): A list of dictionaries, each containing 'read_id',
                    'sample', 'group', and 'sequence'.
      reference (str): The uppercase reference nucleotide sequence.

  Returns:
      list: A new list of dictionaries in the original read order, where each
            dictionary contains 'read_id', 'sample', 'group', normalised
            'sequence', 'status' ('valid' or 'invalid'), 'errors', and
            'mutations'.
  """

  # My code:
  analysed_results = []

  for read in reads:
    # normalise and validate read seq with step 1 func
    validation = validate_read(read["sequence"], reference)
    normalised_sequence = validation["sequence"]

    # call find_mutations on valid reads only
    if validation["valid"]:
      status = "valid"
      mutations = find_mutations(normalised_sequence, reference)
    else:
      status = "invalid"
      mutations = []

    # creating new dictionary as to not change the original
    analysed_reads = {
        "read_id": read["read_id"],
        "sample": read["sample"],
        "group": read["group"],
        "sequence": normalised_sequence,
        "status": status,
        "errors": validation["errors"],
        "mutations": mutations
    }

    analysed_results.append(analysed_reads)

  return analysed_results
````

````python
def get_top_positions(position_counts, max_positions=3):
    """
    Returns up to `max_positions` nucleotide positions with the highest
    mismatch counts, breaking ties by selecting earlier positions first.
    """
    # Sort by highest count descending (-count) then earlier position ascending (pos)
    sorted_positions = sorted(
        position_counts.keys(),
        key=lambda pos: (-position_counts[pos], pos)
    )
    return sorted_positions[:max_positions]

def print_group_summaries(summaries):
    """
    Prints formatted summary and substitution mutation tables separately
    for each experimental group in the sample summaries.
    """
    # Group sample names by their respective experimental group (preserving order)
    groups = {}
    for sample_name, data in summaries.items():
        group_name = data["group"]
        if group_name not in groups:
            groups[group_name] = []
        groups[group_name].append(sample_name)

    # Print tables separately for each group
    for group_name, sample_list in groups.items():
        print("=" * 82)
        print(f"Group Summary: {group_name}")
        print("=" * 82)

        # Table 1: Summarised Results
        print("Summarised results table:")
        header = (
            f"{'sample':<12} | {'valid':>8} | {'invalid':>8} | "
            f"{'mismatch %':>12} | {'events':>8} | {'Top Positions':<18}"
        )
        print(header)
        print("-" * len(header))

        for sample in sample_list:
            s_data = summaries[sample]
            if s_data["mismatch_percentage"] is None:
                pct_str = "N/A (0 valid)"
            else:
                pct_str = f"{s_data['mismatch_percentage']:.2f}%"

            top_pos_str = (
                ", ".join(str(p) for p in s_data["top_positions"])
                if s_data["top_positions"] else "None"
            )

            print(
                f"{sample:<12} | {s_data['valid_reads']:>8} | "
                f"{s_data['invalid_reads']:>8} | {pct_str:>12} | "
                f"{s_data['mismatch_events']:>8} | {top_pos_str:<18}"
            )

        # Table 2: Substitution Mutations
        print("\nTable 2: Substitution Mutations — Count (per 100 valid reads %)")

        # Collect all unique substitutions observed across samples in this group
        observed_subs = sorted({
            sub
            for sample in sample_list
            for sub in summaries[sample]["substitution_counts"].keys()
        })

        if not observed_subs:
            print("No substitution mutations observed in this group.\n")
            continue

        # Build column headers
        sub_header = f"{'substitution':<14}"
        for sample in sample_list:
            sub_header += f" | {sample:>18}"
        print(sub_header)
        print("-" * len(sub_header))

        # Print rows for each observed substitution
        for sub in observed_subs:
            display_sub = sub.replace(">", " -> ")
            row_str = f"{display_sub:<14}"
            for sample in sample_list:
                s_data = summaries[sample]
                count = s_data["substitution_counts"].get(sub, 0)
                valid = s_data["valid_reads"]
                if valid == 0:
                    cell_str = f"{count} (N/A)"
                else:
                    rate_per_100 = (count / valid) * 100
                    cell_str = f"{count} ({rate_per_100:.2f}%)"
                row_str += f" | {cell_str:>18}"
            print(row_str)
        print()

def summarise_samples(analysed_reads):
    """
    Aggregates read validation and mismatch statistics by sample and prints
    group-level summary and substitution tables.
    """
    if not analysed_reads:
        return {}

    summaries = {}

    for read in analysed_reads:
        sample = read["sample"]
        group = read["group"]

        # Initialise sample entry
        if sample not in summaries:
            summaries[sample] = {
                "group": group,
                "valid_reads": 0,
                "invalid_reads": 0,
                "mismatch_reads": 0,
                "mismatch_percentage": None,
                "mismatch_events": 0,
                "substitution_counts": {},
                "position_counts": {},
                "top_positions": []
            }

        sample_summary = summaries[sample]

        # Count valid vs invalid reads
        if read["status"] == "valid":
            sample_summary["valid_reads"] += 1
            mutations = read["mutations"]

            # Count mismatch-containing reads and events
            if len(mutations) > 0:
                sample_summary["mismatch_reads"] += 1
                sample_summary["mismatch_events"] += len(mutations)

                for mut in mutations:
                    sub_key = f"{mut['reference_base']}>{mut['read_base']}"
                    pos_key = mut["position"]

                    sample_summary["substitution_counts"][sub_key] = (
                        sample_summary["substitution_counts"].get(sub_key, 0) + 1
                    )
                    sample_summary["position_counts"][pos_key] = (
                        sample_summary["position_counts"].get(pos_key, 0) + 1
                    )
        else:
            sample_summary["invalid_reads"] += 1

    # Finalise percentages and top positions
    for sample, sample_summary in summaries.items():
        valid_count = sample_summary["valid_reads"]
        if valid_count > 0:
            sample_summary["mismatch_percentage"] = (
                (sample_summary["mismatch_reads"] / valid_count) * 100
            )
        else:
            sample_summary["mismatch_percentage"] = None

        # Calculate top positions regardless of whether count is > 0 (will handle empty list gracefully)
        sample_summary["top_positions"] = get_top_positions(
            sample_summary["position_counts"], max_positions=3
        )

    # Print formatted tables for each experimental group (should be completely outside the loop)
    print_group_summaries(summaries)

    return summaries
````

## Function:

### 👍 You can assume

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

### 🧮 Sample-level calculations

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python

```

Each list must contain **one percentage per biological sample**.

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

````python
# Write your code here.
def calculate_mean(values):
  """
  calculates mean of a number list or returns None if list is empty
  """
  if not values:
    return None
  return sum(values) / len(values)

def combine_position_counts(sample_summary, target_groups):
  """
  sums mismatch counts at each nt position across every valid sample
  in specified treated and control groups
  """
  combined_counts = {}
  for sample_name, data in sample_summary.items():
    if data["group"] in target_groups and data["valid_reads"] > 0:
      for pos, count in data["position_counts"].items():
        if count > 0:
          combined_counts[pos] = combined_counts.get(pos, 0) + count
  return combined_counts


def print_position_line(pos_result, control_group, treated_group):
  """
  prints single summary for a position in scientific style,
  reporting group means, sample sizes, differences and welch's t-test results
  """
  pos = pos_result["position"]
  c_n = pos_result["control_n"]
  t_n = pos_result["treated_n"]

  c_mean_str = (
      f"{pos_result['control_mean']:.2f}%"
      if pos_result["control_mean"] is not None else "N/A"
  )
  t_mean_str = (
      f"{pos_result['treated_mean']:.2f}%"
      if pos_result["treated_mean"] is not None else "N/A"
  )
  diff_str = (
      f"{pos_result['difference']:+.2f} percentage points"
      if pos_result["difference"] is not None else "N/A"
  )
  if pos_result["statistic"] is not None and pos_result["p_value"] is not None:
    t_val = pos_result["statistic"]
    p_val = pos_result["p_value"]
    p_str = f"p = {p_val:.4f}" if p_val >= 0.0001 else "p < 0.0001"
    stats_str = f"two-sided Welch's t-test: t = {t_val:.3f}, {p_str}"
  else:
    stats_str = "two-sided Welch's t-test: t = N/A, p = N/A"


  print(
      f"Position {pos}: {control_group} mean = {c_mean_str} (n={c_n}), "
      f"{treated_group} mean = {t_mean_str} (n={t_n}), "
      f"difference = {diff_str} ({stats_str})"
  )


def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
    """
    identifies top 3 mismatch positions across treated and control groups,
    calulates sample-level mismatch percentages while comparing groups using
    a 2 sided Welch's t-test

    parameters:
       sample_summary (dictionary) - outputs dictionary from 'summarise_samples'
       control_group (str) - name of untreated control group
       treated_group (str) - name of the treated experimental group

    returns:
      list - list of dictionaries containing 'control_n', 'position',
      'treated_n', 'control_mean', 'treated_mean', 'difference', 'statistic' and 'p_value'
    """
    if not sample_summary:
        return []
    target_groups = {control_group, treated_group}

    #filter for valid samples in each target group, reporting any exclusions
    valid_samples = {control_group: [], treated_group: []}
    for sample_name, data in sample_summary.items():
      group = data["group"]
      if group in target_groups:
        if data["valid_reads"] == 0:
          print(
                  f"Excluded sample '{sample_name}' ({group}): "
                  f"0 valid reads available."
              )
        else:
          valid_samples[group].append(sample_name)
    #1. sum mismatch counts across all treated and control groups
    combined_counts = combine_position_counts(sample_summary, target_groups)

    #2. Resuse step 4 helper functions to select the top 3 positions
    top_positions = get_top_positions(combined_counts, max_positions=3)

    #3. calculate means, differences, sample-level percentages and statistics
    results = []

    for pos in top_positions:
      #calculates % of valid reads with a mismatch at 'pos' for control samples
      control_percentages = [
         (sample_summary[s]["position_counts"].get(pos, 0) / sample_summary[s]["valid_reads"]) * 100
         for s in valid_samples[control_group]
      ]

      #calculate percentage of valid reads with a mismatch at 'pos' for treated samples
      treated_percentages = [
          (sample_summary[s]["position_counts"].get(pos, 0) / sample_summary[s]["valid_reads"]) * 100
          for s in valid_samples[treated_group]
      ]

      control_mean = calculate_mean(control_percentages)
      treated_mean = calculate_mean(treated_percentages)

      if control_mean is not None and treated_mean is not None:
        difference = treated_mean - control_mean
      else:
        difference = None

      #Run 2 sided Welch's t-test via given function
      statistical_result = calculate_statistics(control_percentages, treated_percentages)

      pos_result = {
            "position": pos,
            "control_n": len(control_percentages),
            "treated_n": len(treated_percentages),
            "control_mean": control_mean,
            "treated_mean": treated_mean,
            "difference": difference,
            "statistic": statistical_result["statistic"],
            "p_value": statistical_result["p_value"]
        }
      results.append(pos_result)

      #print one summary line for this position
      print_position_line(pos_result, control_group, treated_group)

    return results
````
